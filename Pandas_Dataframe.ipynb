Q10. Aggregation Analysis

Using the Sales column, calculate:

1. Total sales

2. Average sales

3. Maximum sales

4. Minimum sales

```python
# 1. Total Sales
total_sales = df["Sales"].sum()
print("Total Sales",total_sales)
```

```python
# 2. Average Sales
avg_sales = df["Sales"].mean()
print("Average Sales", round(avg_sales,2))
```

```python
# 3. Maximum Sales
max_sales = df["Sales"].max()
print("Max Sales", max_sales)
```

```python
# 4. Min Sales
min_sales = df["Sales"].min()
print("Min Sales", min_sales)
```
